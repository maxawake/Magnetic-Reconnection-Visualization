```python
import numpy as np
import matplotlib.ticker as tk
import sys
sys.path.append("../src/")
import matplotlib.pyplot as plt
from mrvis.utils import matplotlib_style 


fig, ax = plt.subplots(1,2,figsize=(7,3), dpi=150)
xlin = np.linspace(-4, 4, 500)

sech = np.sqrt(1-np.tanh(xlin)**2)

J_num = np.abs(sech)**2
I_2_num = 0.25*(sech)**4

ax[0].plot(xlin, J_num, label=r"$\|\mathbf{J}\|_2$", linestyle="--", color="black")
ax[0].plot(xlin, I_2_num, label="$I_2$", color="black")

# Add labels and title
ax[0].set_xlabel('$y$', fontname='serif')
ax[0].set_title('Side view, Time $t=0$', fontname='serif')
ax[0].legend()

xlin = np.linspace(-9, 9, 500)

sech = np.sqrt(1-np.tanh(xlin)**2)

J_num = np.abs(np.cos(xlin)+1)
I_2_num = 0.25*(np.cos(xlin)-1)**2

ax[1].plot(xlin / np.pi, J_num, label=r"$\|\\mathbf{J}\\|_2$", linestyle="--", color="black")
ax[1].plot(xlin / np.pi, I_2_num, label="$I_2$", color="black")


ax[1].xaxis.set_major_formatter(tk.FormatStrFormatter('%g $\\pi$'))
ax[1].xaxis.set_major_locator(tk.MultipleLocator(base=1.0))

# Add labels and title
ax[1].set_xlabel('$x$')
ax[1].set_title('Top view, Time $t=1$', fontname='serif')

plt.tight_layout()
plt.show()
```

```python
values = []
for i in range(10, 114):
    path = FILE_PATH + f"extraction_3/result_values_{i}.0.txt"
    number = np.loadtxt(path)
    print(i, path)
    values.append([i,number])
    
def rgb_to_hex(rgb):
    hex_code = '#{:02x}{:02x}{:02x}'.format(rgb[0], rgb[1], rgb[2])
    return hex_code

all_hists = []
plt.figure(figsize=(4,3), dpi=150)
for i in range(104):
    hist, bins = np.histogram(np.arccos(np.abs(values[i][1]))/np.pi*180, bins=100)
    all_hists.append(hist)
    plt.plot(bins[:-1], hist, color=rgb_to_hex((int(250-250/104*i),int(250-250/104*i),int(250-250/104*i))), alpha=0.75)
    plt.yscale("log")
    plt.xlim(20,90)
    plt.xlabel("Angle $\\theta$ in $^\circ$")
    plt.ylabel("Number of Occurences")
# plt.savefig("/home/max/Nextcloud/Bachelor/Media/Plots/histograms.pdf", dpi=300)
```

```python
values_total = []
for i in range(0, 114):
    path = f"/run/media/max/Data/Simulations/Plasma/hybrid_3d_turbulence/angle-distribution/result_values_{i}.0.txt"
    number = np.loadtxt(path)
    hist, bins = np.histogram(np.arccos(np.abs(number))/np.pi*180, bins=100)
    print(i, path)
    values_total.append([i,hist])
    
hist_all = np.sum(all_hists, axis=0)
plt.plot(bins[:-1], hist_all, color="black")
plt.yscale("log")

plt.figure(figsize=(5,4))
hist_total = np.sum([values_total[i][1] for i in range(1,104)], axis=0)
plt.plot(bins[:-1], hist_total/np.sum(hist_total), color="Black", label="Whole Domain", linewidth=2)
plt.plot(bins[:-1], hist_all/np.sum(hist_all), color="#743db4", label="Pseudo-Bifurcation Lines", linewidth=2)   
plt.yscale("log")
plt.ylabel("Probability Density")
plt.xlabel(r"Angle $\\theta$ in $^\circ$")
plt.legend()
# plt.savefig("/home/max/Nextcloud/Bachelor/Media/Plots/angle_distribution.pdf", dpi=300)
```
